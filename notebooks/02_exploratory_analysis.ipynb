```python
import sys
from pathlib import Path

# Add src/ directory to python path
sys.path.append(str(Path.cwd().parent / "src"))

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns

from marketlens.returns import (
    calculate_cumulative_returns,
    calculate_normalized_prices,
    calculate_rolling_volatility,
    calculate_simple_returns,
)

# Plotting parameters
sns.set_theme(style="whitegrid", palette="muted")
plt.rcParams["figure.figsize"] = (12, 6)
plt.rcParams["figure.dpi"] = 100

# Load cleaned dataset
prices = pd.read_csv(
    "../data/processed/cleaned_prices.csv", index_col=0, parse_dates=True
)
daily_returns = calculate_simple_returns(prices)

print(f"Prices matrix shape: {prices.shape}")
print(f"Daily returns shape: {daily_returns.shape}")
```

```python
normalized_prices = calculate_normalized_prices(prices, base_value=100.0)

fig, ax = plt.subplots()
normalized_prices.plot(ax=ax, linewidth=1.8)
ax.set_title(
    "Normalized Asset Growth (Base = $100)", fontsize=14, fontweight="bold"
)
ax.set_ylabel("Normalized Price ($)")
ax.set_xlabel("Date")
ax.legend(title="Tickers", frameon=True)
plt.tight_layout()
plt.savefig("../reports/figures/01_normalized_performance.png")
plt.show()
```

```python
fig, ax = plt.subplots(figsize=(8, 6))
corr_matrix = daily_returns.corr()

sns.heatmap(
    corr_matrix,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    vmin=-1,
    vmax=1,
    square=True,
    ax=ax,
)
ax.set_title(
    "Daily Return Correlation Matrix (2019-2024)",
    fontsize=14,
    fontweight="bold",
)
plt.tight_layout()
plt.savefig("../reports/figures/02_correlation_matrix.png")
plt.show()
```

```python
rolling_vol = calculate_rolling_volatility(daily_returns, window=20)

fig, ax = plt.subplots()
rolling_vol.plot(ax=ax, linewidth=1.5)
ax.set_title(
    "20-Day Annualized Rolling Volatility", fontsize=14, fontweight="bold"
)
ax.set_ylabel("Annualized Volatility (σ)")
ax.set_xlabel("Date")
ax.legend(title="Tickers", frameon=True)
plt.tight_layout()
plt.savefig("../reports/figures/03_rolling_volatility.png")
plt.show()
```
